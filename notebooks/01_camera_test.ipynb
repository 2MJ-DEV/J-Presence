# 01 - Test de la camera

Objectif : verifier que la webcam est accessible depuis l'environnement Conda du projet.

Ce notebook teste :
- l'ouverture de la camera ;
- la lecture d'une image ;
- l'affichage d'une capture ;
- la fermeture propre de la camera.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

from src.config.settings import get_settings
from src.camera.webcam import open_camera, read_frame, release_camera

settings = get_settings()
settings.camera_index
```

## Lire une image depuis la webcam

Si cette cellule echoue, verifier d'abord :
- que la webcam n'est pas deja utilisee par une autre application ;
- que `CAMERA_INDEX` vaut bien `0` dans `.env` ;
- que le notebook utilise le kernel `lab-attendance-ai`.

```python
capture = open_camera(settings.camera_index)
try:
    frame = read_frame(capture)
    print(f"Image capturee : {frame.shape}")
finally:
    release_camera(capture)
```

## Afficher la capture

OpenCV lit les images en BGR. Matplotlib attend RGB, donc on convertit les couleurs avant affichage.

```python
import cv2
import matplotlib.pyplot as plt

rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(8, 5))
plt.imshow(rgb_frame)
plt.axis("off")
plt.show()
```

## Test optionnel du flux video

Executer cette cellule seulement sur votre machine locale. Appuyer sur `q` pour fermer la fenetre.

```python
import cv2

capture = open_camera(settings.camera_index)
try:
    while True:
        frame = read_frame(capture)
        cv2.imshow("Lab Attendance AI - camera test", frame)
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break
finally:
    release_camera(capture)
```
